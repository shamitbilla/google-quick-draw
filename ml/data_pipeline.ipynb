```python
import os
import random
import requests
import ndjson
from PIL import Image, ImageDraw
from io import StringIO
from urllib.parse import quote
from concurrent.futures import ThreadPoolExecutor
from threading import local



base_url = "https://storage.googleapis.com/quickdraw_dataset/full/simplified/"
thread_local = local()

def get_session():
    if not hasattr(thread_local, "session"):
        thread_local.session = requests.Session()
    return thread_local.session


def data_pipeline(category):
    try:
        print(f"➡️  Starting: {category}")
        url = base_url + quote(category) + '.ndjson'
        session = get_session()
        response = session.get(url, timeout=10)

        buffer = StringIO(response.text)
        reader = ndjson.reader(buffer)
        drawings = [line for _, line in zip(range(1000), reader) if line["recognized"]]

        if not drawings:
            print(f"⚠️ No recognized drawings for: {category}")
            return category

        sampled_100 = random.sample(drawings, k=min(100, len(drawings)))

        os.makedirs(f"data/{category}", exist_ok=True)

        for count, drawing in enumerate(sampled_100):
            img = Image.new("L", (256, 256), color=255)
            draw = ImageDraw.Draw(img)
            for stroke in drawing['drawing']:
                points = list(zip(stroke[0], stroke[1]))
                draw.line(points, fill=0, width=2)
            img.save(f"data/{category}/_{count}.png")

        print(f"✅ Finished: {category}")
        return category

    except Exception as e:
        print(f"❌ Error in {category}: {e}")
        return category
```

```python
with ThreadPoolExecutor(max_workers=32) as executor:
    futures = {executor.submit(data_pipeline, category): category for category in categories}

    for future in tqdm(as_completed(futures), total=len(futures), desc="Processing Categories: "):
        category_done = future.result()
        tqdm.write(f"✅ Finished: {category_done}")
```
